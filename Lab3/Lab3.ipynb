## Apply descriptive statistical techniques to summarize and interpret datasets

```python
import pandas as pd

columns = [
    "Class",
    "Alcohol",
    "Malic_Acid",
    "Ash",
    "Alcalinity_of_Ash",
    "Magnesium",
    "Total_Phenols",
    "Flavanoids",
    "Nonflavanoid_Phenols",
    "Proanthocyanins",
    "Color_Intensity",
    "Hue",
    "OD280_OD315",
    "Proline"
]

df = pd.read_csv("wine.data", header=None, names=columns)
```

### Mean (Average)

The mean represents the average value of each chemical attribute across all wine samples.

```python
df.mean()
```

### Median (Middle Value)

The median shows the central tendency and is less affected by extreme values compared to the mean.

```python
df.median()
```

### Mode (Most Frequent Value)

The mode represents the most frequently occurring value in each feature. It is useful for identifying common measurements.

```python
df.mode()
```

### Minimum and Maximum

```python
df.min()
```

```python
df.max()
```

### Sum

```python
df.sum()
```

### Standard Deviation (Spread of Data)

Standard deviation measures how much the data varies from the mean. Higher values indicate greater variability among wine samples.

```python
df.std()
```

###  Quartiles & Percentiles

```python
df.quantile([0.25, 0.5, 0.75])
```

```python
df.quantile([0.10, 0.90])
```

### Combined Statistical Summary

```python
df.describe()
```

```python
df.corr()
```

```python
df.cov()
```
